## Beginner — for Loop

### 1. Print numbers from 1 to 10

```python
a = int(input("Enter a starting number:"))
b = int(input("Enter a ending number:"))
for i in range(a , b):
  print(i)
```

### 2. Print numbers from 10 to 1

```python
a = int(input("Enter a starting number:"))
b = int(input("Enter a ending number:"))
for i in range(a , b , -1):
  print(i)
```

### 9. Count how many even numbers are between 1 and 50.

```python
count = 0
for i in range(1,51):
  if i % 2 == 0:
    count = count + 1

print("Total Even Numbers:", count)
```

### 11. Take a number from the user and calculate its factorial.

```python
num = int(input("Enter a number:"))
print("Your number is:", num)

factorial = 1

for i in range(1, num+1):
  factorial = factorial * i

print("Factorial =", factorial)
```

### 14. Count vowels in a string

```python
text = input("Enter a string: ")

count = 0

for character in text:
    if character in "aeiouAEIOU":
        count = count + 1

print("Number of vowels =", count)
```

### 15. Find the largest number in a list using a loop.

```python
numbers = [12, 45, 7, 89, 23]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number =", largest)
```

## Challenging

### 16. Reverse a number using a loop.

```python
num = int(input("Enter a number:"))
print("Your number is:", num)

reverse = 0

while num > 0:
  remainder = num % 10
  reverse = (reverse * 10) + remainder
  num = num // 10

print("Reverse =", reverse)
```

### 17. Count the digits of a number using a loop.

```python
num = int(input("Enter a number: "))

count = 0

while num > 0:
    num = num // 10
    count = count + 1

print("Number of digits =", count)
```

### 20. Create a number guessing game using a while loop.

Requirements:

- Store a secret number, e.g. 7.
- Ask the user to guess the number.
- If the guess is too high, print "Too high".
- If the guess is too low, print "Too low".
- Continue asking until the user guesses correctly.
- When correct, print "Correct!".

```python
secret_number = 7

while True:

    guess = int(input("Guess the number: "))

    if guess > secret_number:
        print("Too high")

    elif guess < secret_number:
        print("Too low")

    else:
        print("Correct!")
        break
```
