## Break Tasks

### 1. Write a for loop from 1 to 10. When the number becomes 5, use break to stop the loop.

```python
for i in range(1,10):
  if i == 5:
    break
  print(i)
```

### 2. Write a loop that checks numbers from 1 to 20. Use break when you find the first even number.

```python
for i in range(1, 20):
  if i % 2 == 0:
    print("Even number is:", i)
    break
```

### 4. Write a loop from 1 to 20. Use continue to skip all even numbers and print only odd numbers.

```python
for i in range(1, 20):
  if i % 2 == 0:
    continue
  print(i)
```

### 5. Create a list:

numbers = [10, -5, 20, -3, 15, -8, 30]

Use a loop and continue to skip negative numbers and print only positive numbers.

```python
numbers = [10, -5, 20, -3, 15, -8, 30]

for i in numbers:
  if i < 0:
    continue
  print(i)
```

### 6. Write a loop from 1 to 30. Use continue to skip numbers that are divisible by 3.

```python
for i in range(1, 30):
  if i % 3 == 0:
    continue
  print(i)
```

## Pass Tasks

### 7. Write a loop from 1 to 10.

- If the number is 5, use pass.
- Otherwise, print the number.

```python
for i in range(1, 10):
  if i == 5:
    pass
  else:
    print(i)
```

### 8. Ask the user to enter a number.

- If the number is positive, print "Positive number".
- If the number is negative, use pass.
- If the number is 0, print "Zero".

```python
num = int(input("Enter a number:"))
print("Number is:", num)

if num > 0:
  print(num, "is a positive number:")
elif num < 0:
  pass
else:
  print(num, "is zero")
```

## Mixed break, continue, and pass

### 9. Write a loop from 1 to 20.

Rules:

- If the number is 7, use pass.
- If the number is even, use continue.
- If the number is 15, use break.
- Otherwise, print the number.

```python
for i in range(1, 20):
  if i == 7:
    pass
  elif i % 2 == 0:
    continue
  elif i == 15:
    break
  else:
    print(i)
```

Create a list:

- marks = [45, 80, -1, 67, 0, 92, 110, 75]

Use a loop with these rules:

- If marks are -1, use continue because the student was absent.
- If marks are greater than 100, use break because the data is invalid.
- If marks are 0, use pass.
- Otherwise, print the marks.

```python
marks = [45, 80, -1, 67, 0, 92, 110, 75]

for mark in marks:

    if mark == -1:
        continue

    elif mark > 100:
        break

    elif mark == 0:
        pass

    else:
        print("Marks:", mark)

print("Checking completed")
```
