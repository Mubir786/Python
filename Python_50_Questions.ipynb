### 50. 🔥 Final Challenge — Student Information & Marks Program

Create a program that asks the user for:

- Name
- Age
- City
- University
- Degree
- English marks
- Math marks
- Computer marks
- Physics marks
- Programming marks

Then display:

- Name: ...
- Age: ...
- City: ...

- Degree: ...
