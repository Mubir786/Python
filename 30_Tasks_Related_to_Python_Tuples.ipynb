### 28. Find the elements that appear in both tuples.

```python
a = (1, 2, 3, 4, 5)
b = (3, 4, 5, 6, 7)

result = tuple(x for x in a if x in b)

print(tuple(dict.fromkeys(result)))
```

### 29. Find the second-largest unique value.

```python
numbers = (10, 50, 30, 50, 40, 20)

unique_numbers = sorted(set(numbers))

if len(unique_numbers) >= 2:
    print(unique_numbers[-2])
else:
    print("No second-largest distinct number")
```

### 30. Store three students' names and marks in a nested tuple. Print each student's information and calculate the average marks.

```python
students = (
    ("Ali", 80),
    ("Ahmed", 90),
    ("Sara", 85)
)

total = 0

for name, marks in students:
    print("Name:", name, "| Marks:", marks)
    total += marks

average = total / len(students)

print("Average marks:", average)
```
