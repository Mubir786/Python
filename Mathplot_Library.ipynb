## Bar Chart

#### Simple bar chart

```python
childern = [10, 20, 40, 10, 30]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.bar(colors, childern)
```

```python
childern = [10, 20, 40, 10, 30]
colors = ['red', 'blue', 'green', 'yellow', 'pink']

plt.bar(colors, childern, color='black')
```

#### horizontal bar chart

```python
plt.barh(colors, childern, color='black')
```

```python
df = pd.read_csv('/content/batsman_season_record.csv')
```

```python
df
```

```python
plt.bar(df['batsman'], df['2015'])
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width=0.2, label='2015', color='red')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2, label='2016', color='green')
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'], width=0.2, label='2017', color='black')

plt.legend(loc='best')
plt.show()
```

#### Stacked Bar chart

```python
plt.bar(df['batsman'], df['2017'], label='2017', color='black')
plt.bar(df['batsman'], df['2016'], label='2016', bottom=df['2017'], color='yellow')
plt.bar(df['batsman'], df['2015'], bottom=(df['2016'] +df['2017']), label='2015', color='green')
plt.legend(loc='best')
plt.show()
```

## Pie Chart

1. Used in Univariate and bivariate analysis of data
2. used in a categorical and numerical type data both

```python
data = [23, 45, 100, 20, 49]

plt.pie(data)
```

```python
data = [23, 45, 100, 20, 49]
subjects = ['eng', 'science', 'maths', 'sst', 'urdu']
plt.pie(data, labels=subjects)
plt.show()
```
