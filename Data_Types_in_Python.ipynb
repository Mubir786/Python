#### List

A List stores multiple items in an ordered collection. You can change, add, or remove items

```python
print([1,2,3,4])
```

#### Tuples

A Tuple stores multiple items in an ordered collection, but you cannot change its items after creating it.

```python
print((1,2,3,4))
```

#### Sets

A Set stores unique values. It automatically removes duplicates and does not guarantee a particular item order.

```python
print({1,2,3,4,5})
```

#### Dictionary

A dictionary is used to store data in key-value pairs. It is changeable

```python
print({
    "Name": "Mubir",
    "Dept": "CS",
    "Gender": "Male",
    "Degree": "MPhil CS"
})
```

- List → collection that you can change
- Tuple → collection that you don't change
- Set → collection of unique values
- Dictionary → key → value pairs
