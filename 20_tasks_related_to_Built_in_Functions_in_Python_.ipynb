## Basic Built-in Functions

### 1. Print()

- Create three variables: name, age, and city.
- Use print() to display all three values in one sentence.

```python
name = str(input("Enter your name: "))
age = int(input("Enter your age: "))
city = str(input("Enter your city: "))

print(f"My name is {name}, I am {age} years old, and I live in {city}.")
```

### 2. input()

- Ask the user for their name and age using input().
- Print a message such as:
- "My name is Ali and I am 23 years old."

```python
name = str(input("Enter your name: "))
age = int(input("Enter your age: "))

print(f"My name is {name} and I am {age} years old.")
```

### 3. len()

- Ask the user to enter their name.
- Use len() to find the number of characters in the name.

```python
name = str(input("Enter your name: "))
print(f"Your name has {len(name)} characters.")
```

### 4. type()

- Create variables containing a string, integer, float, list, and boolean.
- Use type() to display the data type of each variable.

```python
a = "Mubir"
b = 2
c = 4.56
d = [1,2,3,4]
e = True

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

### 5. int()

- Take two numbers from the user using input().
- Convert them into integers using int().
- Print their sum.

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print(f"The sum of {a} and {b} is {a+b}.")
```

### 6. float()

- Ask the user to enter the price of a product.
- Convert the input into a float.
- Print the price.

```python
price = float(input("Enter the price of a product: "))
print(f"The price is {price}")
```

### 7. str()

- Store your age in an integer variable.
- Convert it into a string using str().
- Print a sentence using the converted value.

```python
age = 23

age = str(age)

print(f"I am {age} years old")
```

### 8. abs()

- Ask the user for a number.
- Use abs() to print its absolute value.
- Example: -25 → 25

```python
num = int(input("Enter a number"))
result = abs(num)
print("absolute value =", result)
```

### 9. round()

- Create a variable:

- number = 15.6789
- Use round() to round it to 2 decimal places.

```python
number = 15.6789

result = round(number, 2)

print(result)
```

### 10. pow()
- Ask the user for a base and exponent.
- Use pow() to calculate the result.

Example:

- Base = 2
- Exponent = 5
- Result = 32

```python
base = int(input("Enter base: "))
exponent = int(input("Enter exponent: "))

result = pow(base, exponent)

print("Result =", result)
```

### 11. max()
- Create a list of 5 numbers.
- Use max() to find the largest number.

```python
numbers = [45, 12, 89, 23, 7]

largest = max(numbers)

print("Largest number =", largest)
```

### 12. min()
- Create a list of 5 numbers.
- Use min() to find the smallest number.

```python
numbers = [45, 12, 89, 23, 7]

smallest = min(numbers)

print("Smallest number =", smallest)
```

### 13. sum()
- Create a list containing 10 numbers.
- Use sum() to calculate the total.

```python
numbers = [10, 20, 30, 40, 50]

total = sum(numbers)

print("Total =", total)
```

### 14. orted()

- Create a list:

- numbers = [45, 12, 89, 23, 7, 56]
- Use sorted() to arrange the numbers in ascending order.

```python
numbers = [45, 12, 89, 23, 7, 56]

result = sorted(numbers)

print(result)
```

### 15. reversed()
- Create a list of 5 names.
- Use reversed() to display the names in reverse order.

```python
names = ["Ali", "Ahmed", "Mubir", "Hassan", "Usman"]

result = reversed(names)

print(list(result))
```

### 16. list()

- Create a tuple:

- fruits = ("Apple", "Banana", "Mango")
- Convert the tuple into a list using list().

```python
fruits = ("Apple", "Banana", "Mango")

result = list(fruits)

print(result)
print(type(result))
```

### 17. tuple()
- Create a list of 5 numbers.
- Convert the list into a tuple using tuple().

```python
numbers = [10, 20, 30, 40, 50]

result = tuple(numbers)

print(result)
print(type(result))
```

### 18. set()

- Create a list containing duplicate numbers:

- numbers = [1, 2, 2, 3, 4, 4, 5, 5]
- Use set() to remove the duplicates.

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 5]

result = set(numbers)

print(result)
```

### 19. all() and any()

- Create this list:

- numbers = [10, 20, 30, 40, 50]
- Use all() to check whether all numbers are greater than 5.
- Use any() to check whether at least one number is greater than 45.

```python
numbers = [10, 20, 30, 40, 50]

result1 = all(number > 5 for number in numbers)
result2 = any(number > 45 for number in numbers)

print("Are all numbers greater than 5?", result1)
print("Is any number greater than 45?", result2)
```

### 20. Multiple Built-in Functions
- Ask the user to enter 5 numbers.
- Store them in a list.
- Use len(), sum(), max(), min(), and sorted() to display:
- Number of values
- Total
- Largest number
- Smallest number
- Sorted numbers

```python
numbers = []

for i in range(5):
    number = int(input("Enter a number: "))
    numbers.append(number)

print("Numbers:", numbers)
print("Number of values:", len(numbers))
print("Total:", sum(numbers))
print("Largest number:", max(numbers))
print("Smallest number:", min(numbers))
print("Sorted numbers:", sorted(numbers))
```
