### 18. Find the most expensive item

```python
prices = {
    "apple": 400,
    "mango": 280,
    "banana": 180,
    "orange": 250
}

most_expensive = max(prices, key=prices.get)
print("Most expensive item:", most_expensive)
print("Price:", prices[most_expensive])
```

### 19. Count how many times each character appears in "banana".

```python
text = "banana"
frequency = {}

for char in text:
  frequency[char] = frequency.get(char, 0) + 1

print(frequency)
```

## Intermediate Dictionary Tasks (21–30)

### 21. Find the student with the highest marks

```python
marks = {
    "Ali": 80,
    "Ahmed": 75,
    "Mubir": 90,
    "Hamza": 78
}

top_student = max(marks, key=marks.get)
print("Top student:", top_student)
print("Marks:", marks[top_student])
```

### 22. Find the student with the lowest marks

```python
marks
```

```python
lowest_student = min(marks, key=marks.get)
print("Lowest student:", lowest_student)
print("Marks:", marks[lowest_student])
```

### 23. Create a new dictionary containing products costing more than 100.

```python
products = {
    "book": 500,
    "pen": 50,
    "bag": 2800,
    "pencil": 20,
    "shoes": 2000
}

result = {}
for product, price in products.items():
  if price > 100:
    result[product] = price

print(result)
```

### 24. Create a dictionary of squares using a loop

```python
squares = {}

for number in range(1, 6):
  squares[number] = number ** 2

print(squares)
```

### 25. Use dictionary comprehension to calculate cubes

```python
cubes = {}

for number in range(1, 6):
  cubes[number] = number ** 3

print(cubes)
```

#### OR

```python
cubes = {
    number: number ** 3
    for number in range(1, 6)
}

print(cubes)
```

### 26. Count words in a sentence

```python
sentence = "This is a sample sentence. Count the words in this sentence."
words = sentence.split()

word_count = {}
for word in words:
  word_count[word] = word_count.get(word, 0) + 1

print(word_count)
```

### 29. Group students by grade

```python
students = {
    "Ali": "A",
    "Sara": "B",
    "Ahmed": "A",
    "Hamza": "C",
    "Ayesha": "B"
}

groups = {}

for name, grade in students.items():
    if grade not in groups:
        groups[grade] = []

    groups[grade].append(name)

print(groups)
```

### 30. Build a student marks management system. Create a dictionary and perform several operations on it.

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Hamza": 88,
    "Ayesha": 95
}

# 1. Print all students and marks
print("All students:")
for name, marks in students.items():
    print(name, ":", marks)

# 2. Calculate average marks
average = sum(students.values()) / len(students)
print("\nAverage marks:", average)

# 3. Find highest marks
top_student = max(students, key=students.get)
print("Highest:", top_student, students[top_student])

# 4. Find lowest marks
lowest_student = min(students, key=students.get)
print("Lowest:", lowest_student, students[lowest_student])

# 5. Add a new student
students["Usman"] = 81
print("\nAfter adding Usman:", students)

# 6. Update existing student's marks
students["Ali"] = 90
print("\nAfter updating Ali:", students)

# 7. Remove a student
students.pop("Hamza")
print("\nAfter removing Hamza:", students)
```
