## Lecture 13 of 100 days of Python Programming

## Guessing Game

```python
import random

Guess_gm = random.randint(1, 100)
counter = 1

guess = int(input("Enter your guessing number: "))

while guess != Guess_gm:

    if guess < Guess_gm:
        print("Guess higher")
    else:
        print("Guess lower")

    guess = int(input("Enter your guessing number: "))
    counter += 1
print("Correct Answer! 🎉")
print("You took", counter, "attempts to solve the game")
```

### Static Function of Factorial program through recursion

```python
def fact(num):
  if num == 1:
    return 1
  else:
    return num*fact(num-1)

print(fact(5))
```

### Static Function of Palindrome program through recursion

Screenshot 2026-09-26 135808.png

```python
def palindrome(text):
  if len(text)<=1:
    print("palindrome")
  else:
    if text[0] == text[-1]:
      palindrome(text[1:-1])
    else:
      print("Not a palindrome")

palindrome("madam")
palindrome("karak")
palindrome("abba")
```

### Python Program for Fibonacci Series

```python
import time

def fib(m):

    if m == 0 or m == 1:
        return 1
    else:
        return fib(m-1) + fib(m-2)


start = time.time()

result = fib(24)

end = time.time()

print("Result:", result)
print("Time:", end - start)
```

```python
import threading
import time


class BankAccount:

    def __init__(self, username, password, balance):
        self.username = username
        self.password = password
        self.balance = balance

    def login(self, username, password):
        print("Checking login...")
        time.sleep(2)

        if username == self.username and password == self.password:
            print("Login successful!")
            print("Welcome", self.username)
            print("Balance:", self.balance)
        else:
            print("Invalid username or password")


# Create bank account
account = BankAccount("Mubir", "12345", 50000)

# Get login information
username = input("Enter username: ")
password = input("Enter password: ")

# Create a thread
login_thread = threading.Thread(
    target=account.login,
    args=(username, password)
)

# Start the thread
login_thread.start()

# Wait for the thread to finish
login_thread.join()

print("Program finished.")
```
